```python

def vacuum_cleaner(room):
    position = 0

    while 'D' in room:
        print("Room status:", room)

        if room[position] == 'D':
            print("Cleaning room", position + 1)
            room[position] = 'C'
        else:
            print("Room", position + 1, "is clean")

        position = (position + 1) % len(room)

    print("All rooms are clean!")
    print(room)

room = ['D', 'C', 'D', 'D']
vacuum_cleaner(room)
```
