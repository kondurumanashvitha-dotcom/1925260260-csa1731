```python
from sklearn import svm

# Create dataset
X = [[1, 2], [2, 3], [3, 3],
     [6, 7], [7, 8], [8, 8]]

y = [0, 0, 0, 1, 1, 1]

# Create SVM model
model = svm.SVC(kernel='linear')

# Train the model
model.fit(X, y)

# Predict the class
result = model.predict([[2, 2]])

# Display output
print("Predicted Class:", result[0])
```
