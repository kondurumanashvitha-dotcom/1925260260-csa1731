```python

# Alpha-Beta Pruning

def alphabeta(depth, node, is_max, scores,
              alpha, beta, height):

    if depth == height:
        return scores[node]

    if is_max:
        best = float('-inf')

        for i in range(2):
            value = alphabeta(
                depth + 1,
                node * 2 + i,
                False,
                scores,
                alpha,
                beta,
                height
            )

            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            value = alphabeta(
                depth + 1,
                node * 2 + i,
                True,
                scores,
                alpha,
                beta,
                height
            )

            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                break

        return best


scores = [3, 5, 2, 9, 12, 5, 23, 23]

height = 3

result = alphabeta(
    0, 0, True, scores,
    float('-inf'),
    float('inf'),
    height
)

print("Optimal value:", result)
```
