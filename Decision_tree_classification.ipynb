```python
from sklearn.tree import DecisionTreeClassifier

# Create dataset
X = [[1, 2], [2, 3], [3, 4],
     [6, 7], [7, 8], [8, 9]]

y = [0, 0, 0, 1, 1, 1]

# Create Decision Tree model
model = DecisionTreeClassifier()

# Train the model
model.fit(X, y)

# Predict the class
result = model.predict([[2, 3]])

# Display output
print("Predicted Class:", result[0])
```
