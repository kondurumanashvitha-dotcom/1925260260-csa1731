```python

from collections import deque

def solve():
    start = (3, 3, 0)
    goal = (0, 0, 1)

    moves = [(1, 0), (2, 0), (0, 1),
             (0, 2), (1, 1)]

    q = deque([(start, [])])
    visited = set()

    while q:
        state, path = q.popleft()

        if state == goal:
            for step in path + [state]:
                print(step)
            return

        if state in visited:
            continue

        visited.add(state)
        m, c, boat = state

        for dm, dc in moves:
            if boat == 0:
                nm, nc = m-dm, c-dc
                nb = 1
            else:
                nm, nc = m+dm, c+dc
                nb = 0

            rm, rc = 3-nm, 3-nc

            if 0 <= nm <= 3 and 0 <= nc <= 3:
                if (nm == 0 or nm >= nc) and \
                   (rm == 0 or rm >= rc):
                    q.append(((nm, nc, nb),
                              path + [state]))

solve()

```
