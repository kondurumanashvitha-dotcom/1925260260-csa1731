```python
from sklearn.neighbors import KNeighborsClassifier

# Create dataset
X = [[1, 2], [2, 3], [3, 3],
     [6, 7], [7, 8], [8, 8]]

y = ['A', 'A', 'A', 'B', 'B', 'B']

# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(X, y)

# Predict the class
result = model.predict([[2, 2]])

# Display output
print("Predicted Class:", result[0])
```
